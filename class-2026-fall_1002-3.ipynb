```python
response = requests.get(f'{BASE_URL}pokemon/pikachu')
results = response.json()
```

```python
print(results.keys())
print(results['abilities'][0].keys())
print(results['abilities'][0]['ability'])
pokemon_id = results['id']
```

```python
results = requests.get(f'{BASE_URL}pokemon/{pokemon_id}').json()
print(results.keys())
print(results['moves'][0])
print(results['moves'][0]['move']['name'])
print(results['sprites']['front_default'])
name = results['name']
```

```python
results = requests.get(f'{BASE_URL}pokemon-species/{name}').json()
print(results.keys())
results['flavor_text_entries'][0]
```

```python
results = requests.get(f'{BASE_URL}pokemon/{name}').json()['stats']
results
```

```python
results = requests.get(f'{BASE_URL}type/electric').json()
print(results.keys())
print(results['pokemon'][0]['pokemon']['name'])
pokemon_name = results['pokemon'][0]['pokemon']['name']
```
