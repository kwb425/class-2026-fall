```python
BASE_URL = 'https://pokeapi.co/api/v2/'

def pokemon_info(pokemon_name):
    res = requests.get(f'{BASE_URL}pokemon/{pokemon_name.lower().strip()}')
    if res.status_code == 200:
        data = res.json()
        name = data['name'].capitalize()
        types = ', '.join([t['type']['name'] for t in data['types']])
        height = data['height']
        weight = data['weight']
        info = f'Name: {name}\nType: {types}\nHeight: {height}\nWeight: {weight}'
        return info
    return 'Pokémon not found.'

def pokemon_image(pokemon_name):
    res = requests.get(f'{BASE_URL}pokemon/{pokemon_name.lower().strip()}')
    if res.status_code == 200:
        data = res.json()
        img = data['sprites']['other']['official-artwork']['front_default'] or data['sprites']['front_default']
        return [img] if img else []
    return []
```

```python
pokemon_info('pikachu')
```

```python
pokemon_image('pikachu')
```

```python
# Gradio interface setup
with gr.Blocks() as app:
    with gr.Row():
        pokemon_input = gr.Textbox(label="Enter Pokémon name (e.g., pikachu, charizard)")
        search_button = gr.Button("Search")
    with gr.Row():
        pokemon_info_output = gr.Textbox(label="Pokémon Info")
        pokemon_image_output = gr.Gallery(label="Pokémon Image", show_label=False)

    search_button.click(
        fn=lambda x: (pokemon_info(x), pokemon_image(x)),
        inputs=pokemon_input,
        outputs=[pokemon_info_output, pokemon_image_output]
    )

app.launch(debug=True)
```
